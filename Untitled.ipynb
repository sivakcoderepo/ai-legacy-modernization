````python
# Show agents/*.py
agents_dir = os.path.join(os.getcwd(), 'agents')
print('\n\n== agents/ (python files) ==')
try:
    for fn in sorted(os.listdir(agents_dir)):
        if fn.endswith('.py'):
            p = os.path.join(agents_dir, fn)
            print('\n---', fn, '---')
            try:
                with open(p, 'r', encoding='utf-8') as f:
                    print(f.read())
            except Exception as e:
                print('Could not read', fn, e)
except Exception as e:
    print('Could not list agents directory:', e)
````
